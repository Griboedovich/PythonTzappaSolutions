```python
import numpy as np
import matplotlib.pyplot as plt
import itertools
```

```python
#center = left ^ (center | right)

def rule30(left, center, right):
    return left ^ (center | right)

def apply_rule(rule, center):
    left = np.roll(center, -1)
    right = np.roll(center, 1)
    return rule(left, center, right)
    

def gen_ca(rule, width, height):
    start = np.zeros(width, dtype=np.uint8)
    start[width // 2] = 1
    return list(itertools.accumulate(range(height), lambda r, _: apply_rule(rule30, r), initial=start))

history = gen_ca(rule30, 200, 200)
    
img = np.array(history)
plt.imshow(img)
```

```python
SHIFTS = [xy for xy in itertools.product([-1, 0, 1], repeat=2) if xy != (0,0)]

def neight(grid):
    [np.roll(grid, shift, axis=(0, 1)) for shift in SHIFTS]



grid = np.random.rand(50,50) < 0.2
plt.imshow(grid)
```
